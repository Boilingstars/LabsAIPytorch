При выполнении арифметических операций тип данных автоматически приводится к более общему:

```python
a = torch.IntTensor([1, 2, 6, 8])
a += 5

b = torch.ones(4)
b *= a

# Выбрасывает ошибку
a += b
```

Результат сложения вещественного числа с целочисленным, итоговое значение представляет собой вещественное число. Но тип данных массива a — целочисленный и он не может сохранять вещественные числа. Отсюда и возникает эта ошибка. 

### 2. Тригонометрические и статистические функции и методы

```python
a = torch.linspace(0, torch.pi, 10)

print(a.cos())
```

Очень часто на практике требуется вычислять сумму значений элементов тензоров, их среднее значение, а также находить минимальные и максимальные значения. Для этих целей в PyTorch существуют встроенные методы тензоров, выполняющие эти действия:

```python
x = torch.FloatTensor([1, 4, 3, 7, 10, 8, 14, 21, 20, 23])
y = torch.FloatTensor([4, 1, 6, 9, 13, 11, 16, 19, 15, 22])
```

Сумма:

```python
print(x.sum())
```

По конкретной оси для многомерных массивов:

```python
z = x.view(5, 2)

print(z.sum(dim=0))
print(z.sum(dim=1))
```

Среднее:

```python
x.mean()
```

Максимум значение:

```python
x.max()
```

Если вычислить максимальное значение по оси, получим ``torch.return_types.max()``:

```python
print(z.max(dim=1))
```

То есть, здесь возвращается последовательность из двух элементов. Чтобы выделить только максимальные значения нужно дополнительно указать атрибут values:

```python
print(z.max(dim=1).values)
```

Минимум значения:

```python
x.min()
```

Медиана:

```python
print(torch.median(x))
```

Дисперсия:

```python
print(x.var())
```

СКО:

```python
print(x.std())
```

Коэффициент корреляции Пирсона:

```python
# Каждый тензор в отдельной строке
XY = torch.vstack([x, y])

print(XY)
print(torch.corrcoef(XY))
print(torch.cov(XY))
```
